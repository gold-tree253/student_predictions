<h1>Кластеризация</h1>

### Нормализация

```python
X = df.iloc[:, 2:]
Xcol = X.columns
```

```python
X
```

```python
for col in X.select_dtypes(include=[np.number]).columns:
    X[col].fillna(X[col].median(), inplace=True)
for col in X.select_dtypes(include=['object']).columns:
    X[col].fillna(X[col].mode()[0], inplace=True)
```

```python
X = X.drop(['Attendance (%)', 'Study Hours'], axis=1, errors='ignore')
gender_map = {'Female':0, 'Male':1}
X['Gender'] = X['Gender'].map(gender_map)
parentalsupport_map = {'Low': 0, 'Medium': 1, 'High': 2}
X['ParentalSupport'] = X['ParentalSupport'].map(parentalsupport_map)
X['Online Classes Taken'] = X['Online Classes Taken'].astype(int)
```

```python
mms = MinMaxScaler()
X=pd.DataFrame(mms.fit_transform(X), columns=X.columns)
```

```python
X
```

### Метод главных компонент

```python
principal_df = pd.DataFrame(data = PCA(n_components=2).fit_transform(X))
    
fig = plt.figure(figsize=(10, 10))
sns.scatterplot(x = principal_df.iloc[:,0], 
                y = principal_df.iloc[:,1], 
                s=60)

plt.title('PCA plot in 2D')
plt.xlabel('PC1')
plt.ylabel('PC2');
```

```python
def TSNEplot(pred=None):    
    tsne = TSNE(random_state=1)

    X_tsne = tsne.fit_transform(X)

    plt.figure(figsize=(12, 10))
    plt.scatter(X_tsne[:, 0], 
                X_tsne[:, 1], 
                c = pred, 
                edgecolor="none", 
                alpha=0.7, s=40,
        cmap=plt.cm.get_cmap("flare", 4))

    plt.colorbar()
    plt.title("t-SNE projection");
```

```python
TSNEplot()
```

```python
def PCAplot3D(x_data, y_data=None):
    components = PCA(n_components=3).fit_transform(x_data)

    fig = px.scatter_3d(
        components, 
        x=0, y=1, z=2, 
        color=y_data, 
        size=0.1*np.ones(len(X)), 
        opacity = 1,
        title='PCA plot in 3D',
        labels={'0': 'PC 1', '1': 'PC 2', '2': 'PC 3'},
        width=900, height=900
    )
    fig.show()
```

```python
PCAplot3D(X)
```

### Кластеризация

### K-Means

```python
pca = PCA(n_components=2)

kmeans = KMeans(n_clusters=4, random_state=1, n_init='auto').fit(X)
centroids_pca = pca.fit_transform(kmeans.cluster_centers_)
pred_k = kmeans.labels_

principal_df = pd.DataFrame(data = PCA(n_components=2).fit_transform(X))
```

```python
inertia = []
for k in range(1, 8):
    kmeans = KMeans(n_clusters=k, random_state=1, n_init='auto').fit(X)
    inertia.append(np.sqrt(kmeans.inertia_))
    
plt.plot(range(1, 8), inertia, marker="s")
plt.xlabel("$k$")
plt.ylabel("$J(C_k)$");
```

```python
pca = PCA(n_components=2)


kmeans = KMeans(n_clusters=4, random_state=1, n_init='auto').fit(X)
centroids_pca = pca.fit_transform(kmeans.cluster_centers_)
pred_k = kmeans.labels_

principal_df = pd.DataFrame(data = PCA(n_components=2).fit_transform(X))
```

```python
fig = plt.figure(figsize=(10, 10))
sns.scatterplot(x = principal_df.iloc[:,0], 
                y = principal_df.iloc[:,1], 
                hue = pred_k, 
                palette="flare", 
                s=60)
sns.scatterplot(x = centroids_pca[:,0], 
                y =  centroids_pca[:,1],  
                marker="x", 
                s=450, 
                color = 'blue')
plt.title('PCA plot in 2D')
plt.xlabel('PC1')
plt.ylabel('PC2');
```

```python
TSNEplot(pred_k)
```

```python
PCAplot3D(X, pred_k)
```

### Agglomerative Clustering by Ward

```python
import scipy.cluster.hierarchy as sch
from matplotlib import pyplot
pyplot.figure(figsize=(12, 5))
dendrogram = sch.dendrogram(sch.linkage(X, method = 'ward'))
plt.title('Dendrogram')
plt.ylabel('Euclidean distances')
plt.show()
```

```python
clustering_ward = AgglomerativeClustering(4).fit(X)
pred_ward = clustering_ward.labels_
```

```python
fig = plt.figure(figsize=(10, 10))
sns.scatterplot(x = principal_df.iloc[:,0], 
                y = principal_df.iloc[:,1], 
                hue = pred_ward, 
                palette="flare", 
                s=60)
plt.title('PCA plot in 2D')
plt.xlabel('PC1')
plt.ylabel('PC2');
```

```python
TSNEplot(pred_ward)
```

```python
PCAplot3D(X, pred_ward)
```

### Spectral Clustering

```python
clustering_sp = SpectralClustering(4).fit(X)
pred_sp = clustering_sp.labels_
```

```python
fig = plt.figure(figsize=(10, 10))
sns.scatterplot(x = principal_df.iloc[:,0], 
                y = principal_df.iloc[:,1], 
                hue = pred_sp, 
                palette="flare", 
                s=60)
plt.title('PCA plot in 2D')
plt.xlabel('PC1')
plt.ylabel('PC2');
```

```python
TSNEplot(pred_sp)
```

```python
PCAplot3D(X, pred_sp)
```

```python
X['Cluster'] = pred_sp
```

```python
y = X['Cluster']
del X['Cluster']
```

```python
X
```

<h1>Классификация</h1>

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
print(X_train.shape, X_test.shape, y_train.shape, y_test.shape)
```

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
```

### KNeighborsClassifier (Классификация ближайших соседей)

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
%%time
knn = KNeighborsClassifier()
knn.fit(X_train, y_train)
```

```python
knn.predict(X_test)
```

```python
knn_pred_test = knn.predict(X_test)
display(knn_pred_test[:20], np.array(y_test[:20]))
```

```python
y_train_predict = knn.predict(X_train)
y_test_predict = knn.predict(X_test)

print('train accuracy: ', accuracy_score(y_train, y_train_predict, normalize=True))
print('test accuracy: ', accuracy_score(y_test, y_test_predict, normalize=True))
```

```python
knn_accuracy = accuracy_score(y_test, y_test_predict, normalize=True)
knn_accuracy
```

### Gaussian Naive Bayes (НАИВНЫЙ БАЙЕСОВСКИЙ КЛАССИФИКАТОР)

```python
from sklearn.naive_bayes import GaussianNB
```

```python
%%time

gnb = GaussianNB()
gnb.fit(X, y)
```

```python
gnb_pred_test = gnb.predict(X_test)
display(gnb_pred_test[:20], np.array(y_test[:20]))
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, gnb_pred_test)
```

```python
gnb_accuracy = accuracy_score(y_test, gnb_pred_test)
```

### Random Forest Classifier (Классификатор дерева решений)

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
%%time
rfc  = RandomForestClassifier(criterion = 'gini',
                                    n_estimators = 100,
                                    max_depth = 9,
                                    random_state = 42,
                                    n_jobs = -1)

rfc.fit(X_train, y_train)
y_pred = rfc.predict(X_test)
```

```python
rfc_pred_test = rfc.predict(X_test)
display(rfc_pred_test[:20], np.array(y_test[:20]))
```

```python
from sklearn.metrics import accuracy_score
```

```python
rfc_accuracy = accuracy_score(y_test, rfc_pred_test)
```

### Метрики классификации

```python
from sklearn.metrics import accuracy_score, mean_absolute_error, mean_squared_error, precision_score, recall_score, f1_score, confusion_matrix, r2_score
from sklearn.metrics import classification_report
```

```python
classifiers =["KNeighborsClassifier", "Gaussian Naive Bayes", "Random Forest Classifier"]

log_cols = ["Classifier", "Accuracy"]
log = pd.DataFrame(columns=log_cols)

log["Classifier"] = classifiers
log["Accuracy"] = [knn_accuracy, gnb_accuracy, rfc_accuracy]
```

```python
log
```

```python
plt.xlabel('Accuracy')
plt.title('Classifier Accuracy')

sns.set_color_codes("muted")
sns.barplot(x='Accuracy', y='Classifier', data=log.sort_values(by='Accuracy'), color="b");
```

### Сохранение модели Random Forest Classifier
